```python
from gpt4all import GPT4All as GPT4AllBase

from langchain_community.llms.gpt4all import GPT4All
from langchain.callbacks.streaming_stdout import StreamingStdOutCallbackHandler

from langchain.agents import AgentExecutor, create_react_agent

from langchain import hub

from langchain.prompts import PromptTemplate

from langchain_community.tools import ShellTool

from metAIsploit_assistant.types import BASE_MODELS

from logging import Logger

logger = Logger(__name__)
```

```python
from langchain.pydantic_v1 import BaseModel, Field
from langchain.tools import BaseTool, StructuredTool, tool, Tool

# import pymetasploit3

class MetasploitInput(BaseModel):
    ip: str = Field(description="The ip address to scan.")

# @tool("metasploit-nmap", args_schema=MetasploitInput)
def MetasploitNMAPtool(ip: str):
    """
    Create a Metasploit NMAP scan tool for a given domain,
    used by LangChain ReACT agent.
    """

    return f"NMAP found that the IP address {ip} has the following ports open: 80, 443"
```

```python
# print("examples/models/gpt4all-13b-snoozy-q4_0.gguf")
# metAIsploit-assistant/examples/models/gpt4all-13b-snoozy-q4_0.gguf
# hacker_model = BASE_MODELS.SNOOZY
# verbose=True,
# max_tokens=2048,

# gpus = GPT4AllBase(
#     model_name="gpt4all-13b-snoozy-q4_0.gguf",
#     model_path="/Users/kylepermuy/Development/test-metasploit-llm/metAIsploit-assistant/examples/models",
#     verbose=True)

# logger.info(gpus.list_gpus())

# print(gpus)

llm = GPT4All(
    backend="gptj",
    model="/Users/kylepermuy/Development/test-metasploit-llm/metAIsploit-assistant/examples/models/gpt4all-13b-snoozy-q4_0.gguf",
    verbose=True
)

tools = [
    MetasploitNMAPtool
]

tools = [
    Tool(
        name="metasploit-nmap",
        func=MetasploitNMAPtool,
        description="Create a Metasploit NMAP scan tool for a given domain"
    )]

prompt = hub.pull("hwchase17/react-chat")
```

```python
chat_agent = create_react_agent(llm, tools, prompt)

chat_agent_executor = AgentExecutor(
    agent=chat_agent, 
    tools=tools, 
    handle_parsing_errors=True,
    verbose=True
)

chat_agent_executor.invoke({ 
    "input": "Use the Metasploit NMAP tool to scan the IP address 127.0.0.1 ",
    "chat_history": ""
})
```
